```python
# Install libraries
!pip install diffusers transformers accelerate torch -q

# Import libraries
import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt

# Load model
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")

# Text prompt
prompt = "A futuristic humanoid AI robot walking through Tokyo streets at night, neon lights, realistic reflections, cinematic photography, highly detailed"

# Generate image
image = pipe(
    prompt,
    height=512,
    width=512,
    num_inference_steps=50,
    guidance_scale=7.5
).images[0]

# Save image
image.save("generated_image.png")

# Display image
plt.imshow(image)
plt.axis("off")
plt.show()

print("Image Generated Successfully!")
```
